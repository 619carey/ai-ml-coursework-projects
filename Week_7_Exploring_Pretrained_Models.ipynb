## 3. Comparison with Course Machine Learning Models
* **Course Models (Logistic Regression / Random Forest):** Trained from scratch on structured, tabular, numerical, and categorical features (`tenure`, `Contract`, `MonthlyCharges`) to perform binary classification (`Churn` vs `Retained`).
* **Pretrained Generative/NLP Model (Transformer):** Pre-trained on massive corpora of text data using self-supervised learning and deep neural attention mechanisms (Transformers). It understands contextual language semantics, syntax, and sentiment without requiring feature engineering on tabular data.

## 5. Limitations and Precautions in Engineering Applications
Before deploying text-based or generative models in critical telecommunication or engineering pipelines, several risks must be addressed:
* **Hallucination & Bias:** Large models can misinterpret sarcasm, industry-specific slang, or cultural idioms in text feedback.
* **Data Privacy & Compliance:** Customer support logs often contain sensitive personally identifiable information (PII), requiring strict anonymization before feeding into cloud-hosted APIs.
* **Latency & Compute Cost:** Transformer models require significantly higher inference computing resources compared to lightweight tabular models like Logistic Regression.

```python
!pip install -q transformers torch

try:
    from transformers import pipeline

    print("Loading Hugging Face sentiment analysis pipeline...")
    sentiment_analyzer = pipeline(
        "sentiment-analysis",
        model="distilbert-base-uncased-finetuned-sst-2-english",
    )

    sample_feedback = [
        "I am extremely happy with the fiber internet speeds and customer support!",
        "The monthly charges keep increasing and your service dropped three times this week. I want to cancel.",
        "The service is okay, nothing special.",
    ]

    print("\n--- Model Demonstration Results ---")
    results = sentiment_analyzer(sample_feedback)
    for text, res in zip(sample_feedback, results):
        print(f"Feedback: '{text}'")
        print(
            f"Result:   Label -> {res['label']} (Confidence: {res['score']:.4f})\n"
        )

except Exception as e:
    print(f"Error details: {e}")
```
