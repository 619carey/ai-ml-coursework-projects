**Task:** Temperature Conversion Tool  
**Description:** A Python script that utilizes functions, loops, and conditional logic to convert multiple temperatures from Celsius to Fahrenheit and evaluate them against a set threshold.

```python
def celsius_to_fahrenheit(celsius):
    """Function to convert a temperature from Celsius to Fahrenheit.

    Formula: F = (C * 9/5) + 32
    """
    return (celsius * 9 / 5) + 32


def main():
    THRESHOLD_CELSIUS = 30.0

    print("=== Celsius to Fahrenheit Conversion Tool ===")
    print(f"Current threshold is set to {THRESHOLD_CELSIUS}°C.\n")

    while True:
        try:
            count = int(
                input("How many temperatures would you like to convert? ")
            )
            if count > 0:
                break
            print("Please enter a positive integer greater than 0.")
        except ValueError:
            print("Invalid input. Please enter a valid whole number.")

    for i in range(1, count + 1):
        print(f"\n--- Conversion {i} of {count} ---")

        while True:
            try:
                temp_c = float(input("Enter temperature in °C: "))
                break
            except ValueError:
                print("Invalid input. Please enter a numerical value.")

        temp_f = celsius_to_fahrenheit(temp_c)

        print(f"Result: {temp_c:.2f}°C is equal to {temp_f:.2f}°F")

        if temp_c > THRESHOLD_CELSIUS:
            print(
                f"Status: Above the threshold of {THRESHOLD_CELSIUS}°C — It's hot!"
            )
        elif temp_c < THRESHOLD_CELSIUS:
            print(
                f"Status: Below the threshold of {THRESHOLD_CELSIUS}°C — It's cool!"
            )
        else:
            print(f"Status: Exactly at the threshold of {THRESHOLD_CELSIUS}°C!")

    print("\nAll conversions completed successfully!")


if __name__ == "__main__":
    main()
```
